```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(' ','_')
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
labels=['Young adult','Adult','Middle-aged','senior']
df['age_group']=pd.qcut(df['age'],q=4,labels=labels)
```

```python
df[['age','age_group']].head()
```

```python
frequency_mapping={
    'Fortnightly':14,
    'Weekly':7,
    'Monthly':30,
    'Quarterly':90,
    'Bi-Weekely':14,
    'Annually':365,
    'Every 3 months':90
}
df['current_frequency_days']=df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['current_frequency_days','frequency_of_purchases']].head(10)
```
